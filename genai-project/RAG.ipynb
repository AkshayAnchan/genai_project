```python
from rank_bm25 import BM25Okapi
documents = [
    "Python is a popular programming language.",
    "Machine learning enables computers to learn from data.",
    "Large Language Models are useful for question answering.",
    "BM25 is a lexical search algorithm.",
    "Vector search uses dense embeddings.",
    "Reciprocal Rank Fusion combines multiple ranked lists.",
    "Cross Encoders rerank retrieved documents.",
    "Retrieval Augmented Generation improves LLM accuracy.",
    "Transformers are state-of-the-art deep learning models.",
    "FAISS enables efficient similarity search.",
]
len(documents)
doc_ids = list(range(len(documents)))
print(doc_ids)
print(f"Toatl Documnets : {len(documents)}")
query = "Python Faiss cross encoder"
print(query)
print("____"*40)
tokenixed_doc = [
    doc.lower().split(" ")
    for doc in documents  
]
tokenized_query = query.lower().split(" " )
print(tokenixed_doc)
print("____"*40)
print(tokenized_query)
print("____"*40)
print(tokenixed_doc[0],tokenized_query)
```

```python
import numpy as np
```

```python
bm25 = BM25Okapi(tokenixed_doc)
print(bm25)
scores = bm25.get_scores(tokenized_query)
print(scores)
```

```python
indexes = np.argsort(scores)[::-1]
print(indexes)
```

```python
for ind in indexes:
    if scores[ind] == 0:
        break

    print(scores[ind], ":", documents[ind])
```
