```python
model_name_1 = "llama3.2:1b"
model_name_2 = "deepseek-r1:1.5b"
```

# Greedy Decoding - Deterministic 

  Deterministic -- same input will produce same output 


      Pros : 
          Fast and stable 
          Fully deterministic
          Good for factual and structured tasks
    
      Cons : 
          Repitative Outputs
          No creativity 

```python
from langchain_ollama import OllamaLLM

llm1 = OllamaLLM(model = model_name_1, temperature = 0 )
llm2 = OllamaLLM(model = model_name_2, temperature = 0 )
```

```python
# Model 1 
response1 = llm1.invoke("Tell me a joke")
print(response1)
print("--"*40)

response2 = llm1.invoke("Tell me a joke")
print(response2)
print("--"*40)

response3 = llm1.invoke("Tell me a joke")
print(response3)
print("--"*40)
```

```python
response1 == response2 == response3
```

```python
response1 = llm2.invoke("Tell me a joke")
print(response1)
print("--"*40)

response2 = llm2.invoke("Tell me a joke")
print(response2)
print("--"*40)

response3 = llm2.invoke("Tell me a joke")
print(response3)
print("--"*40)
```

```python
response1 == response2 == response3
```

# Temperature Sampling 
    - Gives Non deterministic behaviour to models output 
    - Range 0-2

    Low Temperature ( T < 1 ): 
        --> Distribution becomes sharper 
        --> Shows some determinstic
        --> High probabilty tokens dominate 

    High Temperature ( T > 1):
        --> Distribution becomes flatter 
        --> More Randomness -- verry less deterministic 
        --> Low probabilty tokens gets more choice 


## Temperature > 1

```python
llm1 = OllamaLLM(model = model_name_1, temperature = 1.6 )
llm2 = OllamaLLM(model = model_name_2, temperature = 1.4 )
```

```python
response1 = llm1.invoke("Tell me a joke")
print(response1)
print("--"*40)

response2 = llm1.invoke("Tell me a joke")
print(response2)
print("--"*40)

response3 = llm1.invoke("Tell me a joke")
print(response3)
print("--"*40)
```

```python
response1 == response2 == response3
```

```python
response1 = llm2.invoke("Tell me a joke")
print(response1)
print("--"*40)

response2 = llm2.invoke("Tell me a joke")
print(response2)
print("--"*40)

response3 = llm2.invoke("Tell me a joke")
print(response3)
print("--"*40)
```

# 3. Top-K Sampling
only samples from top k tokens 

Get top k (probability )--> Randomly select any one of them 

```python
llm1 = OllamaLLM(model = model_name_1, model_kwargs={"top_k" : 20})
llm2 = OllamaLLM(model = model_name_2, model_kwargs={"top_k" : 500} )
```

```python
response1 = llm1.invoke("Tell me a joke")
print(response1)
print("--"*40)

response2 = llm1.invoke("Tell me a joke")
print(response2)
print("--"*40)

response3 = llm1.invoke("Tell me a joke")
print(response3)
print("--"*40)
```

```python
response1 == response2 ==response3
```

```python
response1 = llm2.invoke("Tell me a joke")
print(response1)
print("--"*40)

response2 = llm2.invoke("Tell me a joke")
print(response2)
print("--"*40)

response3 = llm2.invoke("Tell me a joke")
print(response3)
print("--"*40)
```

# 4 - Top P sampling 

```python
llm1 = OllamaLLM(model = model_name_1, model_kwargs={"top_p" : 0.9})
llm2 = OllamaLLM(model = model_name_2, model_kwargs={"top_p" : 0.5} )
```

```python
response1 = llm1.invoke("Tell me a joke")
print(response1)
print("--"*40)

response2 = llm1.invoke("Tell me a joke")
print(response2)
print("--"*40)

response3 = llm1.invoke("Tell me a joke")
print(response3)
print("--"*40)
```

```python
response1 = llm2.invoke("Tell me a joke")
print(response1)
print("--"*40)

response2 = llm2.invoke("Tell me a joke")
print(response2)
print("--"*40)

response3 = llm2.invoke("Tell me a joke")
print(response3)
print("--"*40)
```
