```python
from langchain_ollama import ChatOllama
from langchain_core.messages import HumanMessage, AIMessage

model = ChatOllama(model="llama3.2:1b")

chat_history = []

response1 = model.invoke([
    HumanMessage(content="My name is Akshay")
])

chat_history.append(HumanMessage(content="My name is Akshay"))
chat_history.append(AIMessage(content=response1.content))

response2 = model.invoke(
    chat_history + [
        HumanMessage(content="What is my name?")
    ]
)

print(response2.content)
```

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage
from langchain_core.output_parsers import StrOutputParser

model = ChatOllama(model="llama3.2:1b")

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant."),
    MessagesPlaceholder(variable_name="chat_history"),
    ("human", "{question}")
])

parser = StrOutputParser()

chain = prompt | model | parser

chat_history = [
    HumanMessage(content="My favorite language is Python"),
    AIMessage(content="That's great! Python is powerful.")
]

response = chain.invoke({
    "chat_history": chat_history,
    "question": "What is my favorite language?"
})

print(response)
```
