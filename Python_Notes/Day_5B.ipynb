## String Indexing

    Index Start with zero - 0

1. variable_name[index_num]
2. variable_name[start_index : end_index +1]  condition start_ind < end_inx
3. variable_name[start_index : end_index +1 : step]

### Default Values 
   variable_name[start_index : end_index +1 : step] \
   variable_name[  0   :  endof_str   : 1 ]

```python
x = "Hello"
x[2], x[4]
```

```python
y = "Micro Degree"
y[0 : 5 ], y[6 : 12]
```

```python
y = "Micro DegreDe"
y[0 : 10 : 2]
#McoDge
```

```python
y = "Micro DegreDe"
y[0 : 10 : 3]
#MrDr
```

```python
# Default Step value is 1
y = "Micro Degree"
y[0 : 5] , y[0 : 5 : 1] 
```

```python
# Default start index is 0
y = "Micro Degree"
y[:5] , y[0:5]
```

```python
# Default end index is len of str 
y = "Micro Degree Gen AI Classes "
y[5 :] , y[5 : 28]
```

## String In built functions 

## 1. Case conversion methods

```python
x = "miCro Degree course"

print("Upper  : ", x.upper())
print("lower  : ", x.lower())
print("title  : ", x.title())
print("capitalize   : ", x.capitalize())
print("swapcase   : ", x.swapcase())
```

## 2. Searching Methods 

```python
# find() and rfind() returns index of first occurance of substring 
x = "miCro Degree course Degree"
ind = x.find("Degree")
ind2 = x.find("hello")
ind3 = x.find("course")
print("find() : ",ind, ind2, ind3)

rind = x.rfind("course")
print("rfind() : ",rind)

# index() -- same as find but returns error when no substring found 
print("index() : ",x.index("Degree"))
# print("index() : ", x.index("hello"))  -- gievs error 

print("rindex() : ",x.rindex("Degree"))

## count the occurances 
x = "miCro Degree course Degree"
print("count() : ", x.count("Degree"))


y = "https://www.sample.com"
print("startswith() : ", y.startswith("http"))
print("startswith() : ", y.startswith("httpk"))

z = "sample_file.pdf"
print("endswith() : ", z.endswith("pdf"))
print("endswith() : ", z.endswith("png"))
```

## 3. Vlidations Methods 

```python
amount = "200"
amount2 = "200.2"
print("isalnum() : ", amount.isalnum())
print("isalnum() : ", amount2.isalnum())

name = "suresh"
nam2 = "suresh123"

print("isalpha() : ", name.isalpha())
print("isalpha() : ", nam2.isalpha())
```

## 4. Modification Methods 

```python
txt = "I'm learning python, and i like python "
new_txt = txt.replace("python","GenAI")

# replace
print(f"replace() :\n {txt=} \n {new_txt=} " )

# strip methods 
## strip() -- removes spaces at start and end - by default
s1 = "     hello This is hello sample     "
stripped_str = s1.strip()
print(f"strip() :\n {s1=} \n {stripped_str=} " )

s1 = "hello This is hello sample hello"
stripped_str = s1.strip("hello")
print(f"strip() :\n {s1=} \n {stripped_str=} " )

s1 = "hello This is hello sample"
stripped_str = s1.strip("hello")
print(f"strip() :\n {s1=} \n {stripped_str=} " )


s1 = "hello This is hello sample hello"
stripped_str = s1.lstrip("hello")
print(f"lstrip() :\n {s1=} \n {stripped_str=} " )

file_name = "sample_file.pdf"
stripped_str = file_name.rstrip(".pdf")
print(f"rstrip() :\n {file_name=} \n {stripped_str=} " )
```

```python
s1 = "     hello This is hello sample hello    "
s2 = s1.strip()
s3 = s2.strip("hello")
print(f" {s1=}\n {s2=}\n {s3=}")

s4 = s1.strip().strip("hello")
print(f" {s4=}")
```
