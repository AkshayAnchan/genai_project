```python
| Purpose              | List        | Set         | Dict                | String              |
| -------------------- | ----------- | ----------- | ------------------- | ------------------- |
| Add one item         | `append()`  | `add()`     | `d[key]=value`      | ❌                   |
| Add many items       | `extend()`  | `update()`  | `update()`          | ❌                   |
| Remove item by value | `remove()`  | `remove()`  | `pop(key)`          | ❌                   |
| Remove safely        | ❌           | `discard()` | `pop(key, default)` | ❌                   |
| Remove last/random   | `pop()`     | `pop()`     | `popitem()`         | ❌                   |
| Remove all           | `clear()`   | `clear()`   | `clear()`           | ❌                   |
| Count occurrence     | `count()`   | ❌           | ❌                   | `count()`           |
| Find position        | `index()`   | ❌           | ❌                   | `find()`, `index()` |
| Sort data            | `sort()`    | ❌           | ❌                   | `sorted()`          |
| Reverse data         | `reverse()` | ❌           | ❌                   | `[::-1]`            |
| Copy object          | `copy()`    | `copy()`    | `copy()`            | ❌                   |
| Check existence      | `in`        | `in`        | `in`                | `in`                |
| Length               | `len()`     | `len()`     | `len()`             | `len()`             |
S
```

```python
LIST               SET                DICT

append()     ->    add()          ->   d[key]=value

extend()     ->    update()       ->   update()

remove()     ->    remove()       ->   pop(key)

pop()        ->    pop()          ->   pop()

clear()      ->    clear()        ->   clear()

copy()       ->    copy()         ->   copy()
```
