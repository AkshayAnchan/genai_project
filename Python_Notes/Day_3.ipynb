### 1.1. Mathemetical Operators 

      result = left_operand   <operator> right_operand

```python
x = 10 
y = 5.0

print("X : ",x)
print("Y : ",y)

# Addition 
z = x + y 
print("Addition Result : ",z)

# subtraction  
z = x - y 
print("subtraction Result : ",z)

# multiplications  
z = x * y 
print("multiplications Result : ",z)

y = 6.0
print("Y : ", y)
# division  
z = x / y 
print("division Result : ",z)

# floore division  
z = x // y 
print("floor division Result : ",z)

# modulus   
z = x % y 
print("modulus Result : ",z)

# power   
z = x ** y 
print("power Result : ",z)
```

### 1.1.2 Comparision Operator 

```python
x = 10 
y = 5.0

print("X : ",x)
print("Y : ",y)

## x is equals to y ?  -- Yes /No  -- True / False
z = x == y
print("Equal operator", z)


## x is not equals to y ?  -- Yes /No  -- True / False
z = x != y
print("Not Equal ", z)


## x is greater than y ?  -- Yes /No  -- True / False
z = x > y
print("Greater Than ", z)


## x is smaller than y ?  -- Yes /No  -- True / False
z = x < y
print("Smaller Than", z)

## x is greater than or equals to y ?  -- Yes /No  -- True / False
z = x >= y
print("Greater Than Equal ", z)


## x is smaller than or equals to y ?  -- Yes /No  -- True / False
z = x <= y
print("Smaller Than Equal", z)
```
