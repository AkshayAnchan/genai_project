## and , or 


# and 
    T and T  --> T
    T and F  --> F
    F and T  --> F
    F and F  --> F

# or 
    T or T  --> T 
    T or F  --> T
    F or T  --> T
    F or F  --> F


```python
marks = 78
if marks >= 85 and marks<=100:
    print("Congrats you have passed with distiction",marks)

elif marks >= 60 and marks<=84:
    print("Congrats you have passed with first class ",marks)

elif marks >= 45 and marks<=59:
    print("Congrats you have passed with second class ",marks)
    
elif marks >= 35 and marks<=44:
    print("Congrats you have passed with third class ",marks)
    
elif marks<=34 and marks<=100:
    print("Sorry Try again")
else:
    print("Something went wrong", marks)
```
