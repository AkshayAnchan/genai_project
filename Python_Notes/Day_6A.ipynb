## Adding Elements Methods

```python
# .append(item) -- item will be added to right most position 
x = [1,3,4,5,6]
x.append(7)
x.append("hello")
print(x)

# .extend([item1, item2,..]) -- allitems will be added to right most position
# .extend(list)  -- combines both list items into single list
x.extend(["hey", 22,44,55,66])
print(x)
x.extend(["hey"])
print(x)


# .insert(index,item) -- inserts item into given index position
x.insert(1,2)
print(x)

x.insert(1,2)
print(x)

x.insert(0,"start")
print(x)
```

## Rearranging methods 

```python
x = [1,23,4,5,7,8,89]
# .sort() -- numerical/alpha increasing order 
print(x)
x.sort()
print(x)

x.sort(reverse=True)
print(x)


x = ["Ak" ,"Z", "Aa" ,"C", "B"]
# .sort() -- numerical/alpha increasing order 
print(x)
x.sort()
print(x)

x.sort(reverse=True)
print(x)
```

```python
x = ["Ak" ,"Z", "Aa" ,"C", "B", 123]
x.sort()
```

```python
## .reverse()
x = ["Ak" ,"Z", "Aa" ,"C", "B"]
x.reverse()
print(x)
print(x[::-1])
```

## Searching & counting

```python
#.index(item) -- returns idex of item
x = ["Ak" ,"Z", "Aa" ,"C", "B", 123]
ind = x.index("Z")
print(x, ind, x[ind])

ind = x.index("ZHKJ")
print(x, ind, x[ind])
```

```python
#.count(item) -- returns count of items 
x = [1,"hello", 3,4,"hello", 5, "hello",6]
cnt = x.count("hello")
print(x, cnt)
```
