### 1. loop through itterables [list /tuple/dictionary/set]
    Syntax : for var_name in itterable :

```python
x = [1,20,34,"Hello","World",5,6,7,"Done",8,9]
print("Number ", x[0])
print("Number ", x[1])
print("Number ", x[2])
print("Number ", x[3])
print("Number ", x[4])
print("Number ", x[5])
print("Number ", x[6])
print("Number ", x[7])
```

```python
x = [0,"Student", 1,20,34,"Hello","World",5,6,7,"Done",8,9,0]
for i in x:
    print("Number", i)

print("Loop Completed")
```

```python
x = [0,"Student", 1,20,34,"Hello","World",5,6,7,"Done",8,9,0]
for i in x[3:]:
    print("Number", i)

print("Loop Completed")
```

```python
x = [0,"Student", 1,20,34,"Hello","World",5,6,7,"Done",8,9,0]
for i in x[::2]:
    print("Number", i)
print("Loop Completed")
```
