# List Comprehensation
1. newlist = [expression for item in list]
2. newlist = [expression for item in list if condition ]
3. newlist = [expressionforTrue if condition else expressionforFalse for item in list]

newlist = list / set / dictionary \
list = list / tuple /set / dictionary


```python
# squares of given list
x = [1,2,3,4,5,6]
sqr = []
for item in x:
    sqr.append(item**2)

print(sqr)
```

```python
sqr = [item**2 for item in x]
print(sqr)
```

```python
sqr = {item**2 for item in x}
print(sqr)
```

```python
# list comprehensation with if condition
# squares of only even number given list
x = [1,2,3,4,5,6]
sqr = []
for item in x:
    if item%2 == 0:
        sqr.append(item**2)
print(sqr)
```

```python
sqr = [item**2 for item in x if item%2 == 0]
sqr
```

```python
# list comprehensation with if else condition
# show string odd , even
x = [1,2,3,4,5,6]
numtypes = []
for item in x:
    if item%2 == 0:
        numtypes.append("Even")
    else:
        numtypes.append("Odd")
print(numtypes)
```

```python
numtypes = ["EVEN" if item%2 == 0 else "Odd" for item in x] 
numtypes
```

```python
def make_capital(item):
    return item.upper()

strs = [make_capital(item) for item in numtypes]
strs
```

```python
x = (1,2,3,4,5,6)
y = [i**2 for i in x]
y
```

```python
x = {1,2,3,4,5,6}
y = [i**2 for i in x]
y
```

```python
x = {"a":1,"b":2,"c":3}
y = [v**2 for k,v in x.items()]
y
```

```python
x = {"a":1,"b":2,"c":3}
y = {k : v**2 for k,v in x.items()}
y
```
