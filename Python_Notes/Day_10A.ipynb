## Copying dictionary and list 

```python
x = 10 
y = x 
y = 40 
print(x,y)
```

```python
x = [1,2,3,4]
y = x 
y.append("Hello")
print(x,y)
```

```python
x = {"a":20, "b" :40}
y = x 
y["c"] = 90
print(x,y)
```

```python
x = [1,2,3,4]
y = x.copy()
y.append("Hello")
print(x,y)
```

```python
x = {"a":20, "b" :40}
y = x.copy() 
y["c"] = 90
print(x,y)
```

```python
x = [1,2,3,4]
y = x
z = y 
z.append("Hello")
print(x,y,z)
```

## Type Casting

```python
x = 4564
y = str(x)
print(y , type(y))
```

```python
x = 4564
y = str(x)
y = int(y)
print(y , type(y))
```

```python
x = 4564
y = str(x)
y = float(y)
print(y , type(y))
```

```python
x = "3457345s"
y = float(x)
print(y , type(y))
```

```python
x = 2352.457.4568
z = int(x)
z
```

```python
# list to set 
x = [1,2,3,4,5,7,8,2,3,4,5]
y = set(x)
y = list(y)
print(y)
```

```python
x = [1,2,3,4,5,7,8,2,3,4,5]
y = list(set(x))
print(y)
```

```python
# list to tuple
x = [1,2,3,4,5,7,8,2,3,4,5]
x = tuple(x)
print(x)
```
