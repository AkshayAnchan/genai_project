# String


## String creation methods
    1. within " -> "Hello World"
    2. within ' -> 'Hello World'
    3. within """  -> """ This is multiline string"
    4. within '''  -> ''' This is another method for long string '''
    

```python
x = "hello world"
type(x)
```

```python
x = "hello \" world"
type(x), x 
```

```python
x = 'hello \' world'
type(x), x 
```

```python
x = """ this is verry 
  long sample text 
  for the llm call 
  input data """

type(x), x 
```

```python
x = ''' this is verry 
  long sample text 
  for the llm call 
  input data '''

type(x), x 
```

```python
y = "456"
z = '23.78'

k = 567
l = 23.78
type(y), type(z), type(k) , type(l)
```

```python
x = 100
print(x, type(x))

x = "Hello"
print(x, type(x))
```

## String Indexing

    Index Start with zero - 0

1. variable_name[index_num]
2. variable_name[start_index : end_index +1]  condition start_ind < end_inx
3. variable_name[start_index : end_index +1 : step]

### Default Values 
   variable_name[start_index : end_index +1 : step] \
   variable_name[  0   :  endof_str   : 1 ]

```python
x = "Hello"
x[2], x[4]
```

```python
y = "Micro Degree"
y[0 : 5 ], y[6 : 12]
```

```python
y = "Micro DegreDe"
y[0 : 10 : 2]
#McoDge
```

```python
y = "Micro DegreDe"
y[0 : 10 : 3]
#MrDr
```

```python
# Default Step value is 1
y = "Micro Degree"
y[0 : 5] , y[0 : 5 : 1] 
```

```python
# Default start index is 0
y = "Micro Degree"
y[:5] , y[0:5]
```

```python
# Default end index is len of str 
y = "Micro Degree Gen AI Classes "
y[5 :] , y[5 : 28]
```

## Negative indexing 

 x    = H  E  L  L  O \
 +ve  = 0  1  2  3  4 \
 -ve  = -5 -4 -3 -2 -1

```python
x = "Hello"
x[1] , x[-4]
```

```python
x = "Hello"
x[1 : 4] , x[-4 : -1]
```

```python
x = "This is the long sentence example"
```

```python
x[:3] , x[-3:]
```

```python
x = "Hello"
x[::-1]
```

```python
# palindrome 
x = "gadag"
rev_x = x[::-1]
print(x, rev_x)

x == rev_x
```
