```python
| Method         | Purpose                     | Example                                  | Output              | Common Use Case             |
| -------------- | --------------------------- | ---------------------------------------- | ------------------- | --------------------------- |
| `upper()`      | Convert to uppercase        | `"hello".upper()`                        | `"HELLO"`           | Case-insensitive comparison |
| `lower()`      | Convert to lowercase        | `"HELLO".lower()`                        | `"hello"`           | Email/login validation      |
| `capitalize()` | First letter uppercase      | `"hello world".capitalize()`             | `"Hello world"`     | Formatting names            |
| `title()`      | Every word starts uppercase | `"hello world".title()`                  | `"Hello World"`     | Display names, titles       |
| `swapcase()`   | Reverse case                | `"HeLLo".swapcase()`                     | `"hEllO"`           | Text transformations        |
| `find()`       | First occurrence index      | `"banana".find("a")`                     | `1`                 | Search operations           |
| `rfind()`      | Last occurrence index       | `"banana".rfind("a")`                    | `5`                 | Find last match             |
| `count()`      | Count occurrences           | `"banana".count("a")`                    | `3`                 | Frequency analysis          |
| `replace()`    | Replace text                | `"I love Java".replace("Java","Python")` | `"I love Python"`   | Data cleaning               |
| `strip()`      | Remove spaces both sides    | `" hello ".strip()`                      | `"hello"`           | Clean user input            |
| `lstrip()`     | Remove left spaces          | `" hello".lstrip()`                      | `"hello"`           | Remove indentation          |
| `rstrip()`     | Remove right spaces         | `"hello ".rstrip()`                      | `"hello"`           | Clean file data             |
| `split()`      | String → List               | `"a,b,c".split(",")`                     | `['a','b','c']`     | CSV parsing                 |
| `join()`       | List → String               | `",".join(['a','b'])`                    | `"a,b"`             | CSV generation              |
| `startswith()` | Check beginning             | `"python".startswith("py")`              | `True`              | URL/file validation         |
| `endswith()`   | Check ending                | `"resume.pdf".endswith(".pdf")`          | `True`              | File validation             |
| `isalpha()`    | Only letters                | `"Python".isalpha()`                     | `True`              | Name validation             |
| `isdigit()`    | Only digits                 | `"123".isdigit()`                        | `True`              | Age/mobile validation       |
| `isnumeric()`  | Numeric characters          | `"123".isnumeric()`                      | `True`              | Number validation           |
| `isalnum()`    | Letters and digits          | `"Python3".isalnum()`                    | `True`              | Username validation         |
| `islower()`    | All lowercase               | `"hello".islower()`                      | `True`              | Text validation             |
| `isupper()`    | All uppercase               | `"HELLO".isupper()`                      | `True`              | Text validation             |
| `isspace()`    | Only spaces                 | `"   ".isspace()`                        | `True`              | Blank input check           |
| `istitle()`    | Title case                  | `"Hello World".istitle()`                | `True`              | Name/title formatting       |
| `center()`     | Center align                | `"Hi".center(10)`                        | `'    Hi    '`      | Reports                     |
| `ljust()`      | Left align                  | `"Hi".ljust(10)`                         | `'Hi        '`      | Table formatting            |
| `rjust()`      | Right align                 | `"Hi".rjust(10)`                         | `'        Hi'`      | Table formatting            |
| `zfill()`      | Pad zeros                   | `"123".zfill(5)`                         | `"00123"`           | IDs, invoice numbers        |
| `partition()`  | Split into 3 parts          | `"a@b.com".partition("@")`               | `('a','@','b.com')` | Email parsing               |
```
