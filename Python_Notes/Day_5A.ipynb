# List 
    - create using [] seperate items with ,
    - heterogenous : list can have **any data type** int float str bool
    - Mutable values of single item can be changed

```python
# List 
name1 = "student1"
name2 = "student2"
name3 = "student3"

names = "arun varun dinesh"

names = ["student1", "student2", "student3", 40 , 500.034, True , False,]
names
```

```python
# indexing 
names[0:2], names[-1]
```

```python
# mutable
names = ["student1", "student2", "student3", 40 , 500.034, True , False,]
names[0] = "Dinesh"
names[3] = "Sudesh"
print(names)
```

```python
# immutable
s = "name"
s[0] = "N"
```
