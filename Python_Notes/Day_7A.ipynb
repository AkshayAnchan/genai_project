## Modifications

## Adding new items

```python
# add new item 
x = {1,2,3,4}
x.add(5)
print(x)

# supports only immutable 
x.add("hello")
print(x)

x.add((7,8))
print(x)

x.add([7,8])
print(x)
```

```python
# add mutiple new item 
#.update(itterable)
x = {1,2,3,4}
x.update([1,2,3,4,5])
print(x)

x.update((1,2,3,4,5))
print(x)

x.update({1,2,3,4,5})
print(x)
```

## Removing items / deletion 

```python
#.remove(item) removes item if existgs else keyError
x = {1,2,3,4}
x.remove(4)
print(x)

x.remove(78)
print(x)
```

```python
#.discard(item) removes item no error if item does not exist
x = {1,2,3,4}
x.discard(4)
print(x)

x.discard(78)
print(x)
```

```python
#.pop(item) removes arbitory / random value
x = {1,2,3,4}
x.pop()
print(x)

x = {"hello",1,2,3,4}
x.pop()
print(x)
```

```python
#.clear() removes arbitory / random value
x = {1,2,3,4}
x.clear()
print(x)
```

## Set operation methods

```python
a = {1,2,3,4}
b = {3,4,5,6}
```

```python
# unioin - both combined 
c = a.union(b)
print(c)

c = a | b
print(c)
```

```python
# intersection - common items in A and B 
c = a.intersection(b)
print(c)

c = a & b
print(c)
```

```python
# difference - Elements in A but not in B
c = a.difference(b)
print(c)

c = a - b
print(c)

c = b.difference(a)
print(c)

c = b-a
print(c)
```

```python
# symetric_difference -- item exist only in one set 
a = {1,2,3,4,8}
b = {3,4,5,6,8}

c = a.symmetric_difference(b)
print(c)

c = a ^ b
print(c)
```

## relationship / comparision methods

```python
a = {1,2,3,4,8}
b = {3,4,8}
c = {1,2,3,4,5,8}

# a.issubset(b) -- is a subset of b
print(a.issubset(b))
print(b.issubset(a))

z = a <= b
print(z)

z = b <= a
print(z)
```

```python
a = {1,2,3,4,8}
b = {3,4,8}

#.isdisjoin()  -- no common elements in both 
print(a.isdisjoint(b))


a = {1,2}
b = {3,4,8}

#.isdisjoin()  -- no common elements in both 
print(a.isdisjoint(b))
```
