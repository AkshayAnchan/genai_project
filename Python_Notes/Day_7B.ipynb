# dictionary 
  - key : value pair
  - { key : value ,  key : value}
  - key conditions \
        - all keys must be unique else overrides lates value \
        - key can be of any hashable / immutable data types
  - value conditions \
        - can any data type \
        - duplicates \

  - Mutable 

```python
x = {"name" : "Mahesh"}
print(type(x), x)
```

## Key conditions

```python
#- all keys must be unique
x = {"name" : "mahesh", "class" : "10th", "marks" :67, "marks" : 89}
print(x)
```

```python
# - key can be of any hashable / immutable data types
x = { 1 : 567.798,         # int
      576.78 : "world" , # float 
      "hello" : 687.798, # str 
      False :  True,  # bool 
      (1,2,3) : "Output"  # tuple
    }
print(x)

# list is mutable so it will through error 
x = {[1,2,3,4] : "hello"}

      
```

## Mutable - Modifications

```python
# adding item 
x = {"name" : "mahesh"}
x["class"] = 10
x["marks"] = [79,89,56,34,67,78]
print(x)

x = {"name" : "mahesh"}
x.update({"class" :10, "marks" : [79,89,56,34,67,78]})
print(x)
```

```python
x = {"name" : "mahesh"}
x["class"] = ""
print(x)
```

## Accesing items 

```python
x = {'name': 'mahesh', 'class': 10, 'marks': [79, 89, 56, 34, 67, 78]}

# dic[key]

val = x["name"]
print(val)

val = x["class"]
print(val)

val = x["marks"]
print(val)

val = x["percentage"]
print(val)
```

```python
x = {'name': 'mahesh', 'class': 10, 'marks': [79, 89, 56, 34, 67, 78]}

# dict.get(key,default_value)
val = x.get("name")
print(val)

val = x.get("percentage")
print(val)

val = x.get("percentage", "This field is not avialble")
print(val)
```

```python
# Get all keys 
x = {'name': 'mahesh', 'class': 10, 'marks': [79, 89, 56, 34, 67, 78]}

# x.keys() -- returns tuple of keys 
xkeys = x.keys()
print(xkeys)

# x.values() -- returns tuple of values
xvalues = x.values()
print(xvalues)

# x.items()  -- returns pair wise tuple (key, value)
xitems = x.items()
print(xitems)
```
