### 1. loop through itterables [list /tuple/dictionary/set]
    Syntax : for var_name in itterable :

```python
x = [1,20,34,"Hello","World",5,6,7,"Done",8,9]
print("Number ", x[0])
print("Number ", x[1])
print("Number ", x[2])
print("Number ", x[3])
print("Number ", x[4])
print("Number ", x[5])
print("Number ", x[6])
print("Number ", x[7])
```

```python
x = [0,"Student", 1,20,34,"Hello","World",5,6,7,"Done",8,9,0]
for i in x:
    print("Number", i)

print("Loop Completed")
```

```python
x = [0,"Student", 1,20,34,"Hello","World",5,6,7,"Done",8,9,0]
for i in x[3:]:
    print("Number", i)

print("Loop Completed")
```

```python
x = [0,"Student", 1,20,34,"Hello","World",5,6,7,"Done",8,9,0]
for i in x[::2]:
    print("Number", i)
print("Loop Completed")
```

# break , continue , pass

## break

 break a loop once a condition occures 

```python
coins = [1,2,3,4,5,6,7,8,2,10]
collect = 20
collected = 0
for c in coins:
    collected = collected + c
    if collected >=collect:
        break
print(collected)
```

```python
while True:
    play = input("Enter Y to continue N to stop :"  )
    if play.lower() == "n":
        print("User entered n exiting the game")
        break
    print("Playing")

print("Thanks for palying")
```

## continue

 do nothing once a condition occures / skip operations 

```python
orders = ["order1", "order2", "order3" ,"cancled", "order4", "order5"]
for order in orders:
    if order == "cancled":
        break
    print("Processing ...", order)
```

```python
orders = ["order1", "order2", "order3" ,"cancled", "order4", "order5","cancled"]
for order in orders:
    if order == "cancled":
        continue
    print("Processing ...", order)
```

```python
orders = ["order1", "order2", "order3" ,"cancled", "order4", "order5","cancled"]
for order in orders:
    if order == "cancled":
        print("Sorry your order is canceled")
        continue
    
    print("Processing ...", order)
```

```python
orders = ["order1", "order2", "order3" ,"cancled", "order4", "order5","cancled"]
for item in orders:
    print(item)
    break
    print("Processing : ",item)
print("Processing is done")
```

```python
orders = ["order1", "order2", "order3" ,"cancled", "order4", "order5","cancled"]
for item in orders:
    print(item)
    continue
    print("Processing : ",item)
print("Processing is done")
```
