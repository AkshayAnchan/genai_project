```python
| Method / Operation | Purpose                 | Example                | Output      | Common Use Case           |
| ------------------ | ----------------------- | ---------------------- | ----------- | ------------------------- |
| `append()`         | Add element at end      | `[1,2].append(3)`      | `[1,2,3]`   | Add item                  |
| `extend()`         | Add multiple elements   | `[1,2].extend([3,4])`  | `[1,2,3,4]` | Merge lists               |
| `insert()`         | Insert at position      | `[1,2].insert(1,99)`   | `[1,99,2]`  | Add at index              |
| `remove()`         | Remove first occurrence | `[1,2,3].remove(2)`    | `[1,3]`     | Delete value              |
| `pop()`            | Remove by index         | `[1,2,3].pop()`        | `3`         | Remove last item          |
| `clear()`          | Remove all elements     | `[1,2].clear()`        | `[]`        | Reset list                |
| `index()`          | Find position           | `[10,20,30].index(20)` | `1`         | Search element            |
| `count()`          | Count occurrences       | `[1,2,2,3].count(2)`   | `2`         | Frequency count           |
| `sort()`           | Sort list               | `[3,1,2].sort()`       | `[1,2,3]`   | Sorting                   |
| `reverse()`        | Reverse list            | `[1,2,3].reverse()`    | `[3,2,1]`   | Reverse order             |
| `copy()`           | Create copy             | `x.copy()`             | New list    | Backup data               |
| `len()`            | Length                  | `len([1,2,3])`         | `3`         | Count elements            |
| `sum()`            | Total                   | `sum([1,2,3])`         | `6`         | Calculations              |
| `min()`            | Smallest value          | `min([5,2,9])`         | `2`         | Statistics                |
| `max()`            | Largest value           | `max([5,2,9])`         | `9`         | Statistics                |
| `sorted()`         | Return sorted list      | `sorted([3,1,2])`      | `[1,2,3]`   | Sorting without modifying |
| `in`               | Membership              | `2 in [1,2,3]`         | `True`      | Validation                |
| `not in`           | Membership              | `5 not in [1,2,3]`     | `True`      | Validation                |
```
