```python
| Method / Operation | Purpose                     | Example                 | Output          | Common Use Case    |
| ------------------ | --------------------------- | ----------------------- | --------------- | ------------------ |
| `count()`          | Count occurrences           | `(1,2,3,2).count(2)`    | `2`             | Frequency counting |
| `index()`          | Find first occurrence index | `(10,20,30).index(20)`  | `1`             | Search position    |
| `len()`            | Length of tuple             | `len((10,20,30))`       | `3`             | Count elements     |
| `in`               | Membership check            | `20 in (10,20,30)`      | `True`          | Validation         |
| `not in`           | Membership check            | `40 not in (10,20,30)`  | `True`          | Validation         |
| `+`                | Concatenate tuples          | `(1,2)+(3,4)`           | `(1,2,3,4)`     | Combine data       |
| `*`                | Repeat tuple                | `(1,2)*3`               | `(1,2,1,2,1,2)` | Pattern generation |
| `[]`               | Indexing                    | `(10,20,30)[1]`         | `20`            | Access element     |
| `[-1]`             | Negative indexing           | `(10,20,30)[-1]`        | `30`            | Last element       |
| `[start:end]`      | Slicing                     | `(10,20,30,40)[1:3]`    | `(20,30)`       | Extract data       |
| `[::2]`            | Step slicing                | `(10,20,30,40,50)[::2]` | `(10,30,50)`    | Alternate values   |
| `[::-1]`           | Reverse tuple               | `(10,20,30)[::-1]`      | `(30,20,10)`    | Reverse data       |
| `min()`            | Smallest value              | `min((10,20,5))`        | `5`             | Statistics         |
| `max()`            | Largest value               | `max((10,20,5))`        | `20`            | Statistics         |
| `sum()`            | Sum values                  | `sum((10,20,30))`       | `60`            | Calculations       |
| `sorted()`         | Sort values                 | `sorted((30,10,20))`    | `[10,20,30]`    | Sorting            |
| `tuple()`          | Convert to tuple            | `tuple([1,2,3])`        | `(1,2,3)`       | Type conversion    |
```
