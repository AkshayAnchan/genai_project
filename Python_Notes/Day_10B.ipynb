# Function Defination and Calling

```python
def say_hello():
    print("Hello")
    print("Function")
    
```

# Function with single argument 

```python
def say_hello(name):
    print("Hello",name)
    print("Function")
```

```python
say_hello("Suresh")
say_hello()
```

# Function with mutiple arguments

```python
def add(a,b):
    c = a+b
    print("Total is :", c)
```

```python
add(12,12)
```

```python
def add(a,b,c):
    print(f"{a=} {b=} {c=}")
    c = a+b
    print("Total is :", c)
    
```

```python
add(11,22,3)
```

```python
add(11,22)
```

# function with default value arguments 

```python
def add(a,b,c=0):
    print(f"{a=} {b=} {c=}")
    c = a+b
    print("Total is :", c)
```

```python
add(11,22,33)
```

```python
add(11,22)
```

# keyword argument

    - always positional arguments first then keyword arguments

```python
def linear(m,x ,c):
    print(f" {m=} {x=} {c=} ")
    y = m*x + c
    print(y)
```

```python
linear(11,10,22)
```

```python
linear(11,22,10)
```

```python
linear(m=11,c=22,x=10)
```

```python
linear(11,10,c=22)
```

```python
linear(m=22,12,45)
```

## Function return 

```python
def add(a,b,c=0):
    print(f"{a=} {b=} {c=}")
    res = a+b+c
    print("Total is :", res)

result = add(11,22,33)
print(result)
```

```python
def add(a,b,c=0):
    print(f"{a=} {b=} {c=}")
    res = a+b+c
    print("Total is :", res)
    return res
```

```python
result = add(11,22,33)
```

```python
result
```

```python
def add(a,b,c=0):
    print(f"{a=} {b=} {c=}")
    res = a+b+c
    print("Total is :", res)
    return res,a,b,c
```

```python
result = add(11,22,33)
result
```

```python
result, ain, inb , inc  = add(11,22,33)
```

```python
print(result, ain , inb , inc)
```
